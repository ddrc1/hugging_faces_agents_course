```python
import json
import pandas as pd
```

```python
file_path = "./GAIA/2023/validation/metadata.jsonl"
questions: list[dict] = []
with open(file_path, "r") as f:
    for line in f:
        linha_limpa = line.strip()
        if linha_limpa:
            questions.append(json.loads(linha_limpa))
```

```python
df_question = pd.DataFrame(questions)
df_question
```

```python
df_question[["Level", "task_id"]].groupby(by="Level").count()
```

```python
metadata_list = df_question[df_question["Level"] == 1]["Annotator Metadata"].tolist()
```

```python
tools_needed = []
for metadata in metadata_list:
    if not metadata["Tools"] in ["None", "No tools required"]:
        tools = metadata["Tools"].split("\n")
        try:
            tools = [tool.split(". ")[1].replace(".", "").replace("(Optional)", "").strip() for tool in tools]
        except:
            pass
        
        tools_needed.extend(tools)

tools_needed = set(tools_needed)
```
